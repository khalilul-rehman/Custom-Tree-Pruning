```python
import numpy as np
import matplotlib.pyplot as plt

def plot_temp(dataFrame: pd.DataFrame):
    depths = dataFrame["max_depth"].unique()
    # Change ncols to 3
    fig, ax = plt.subplots(nrows=len(depths), ncols=3, figsize=(22, 7 * len(depths)))
    
    if len(depths) == 1:
        ax = np.atleast_2d(ax)
        
    for index, depth in enumerate(depths):
        sub_df = dataFrame[dataFrame["max_depth"] == depth].copy()
        alphas = sub_df["alpha"].astype(str).tolist()
        x = np.arange(len(alphas))
        total_width = 0.8
        
        # --- SUBPLOT 0: Epsilon (Linear) ---
        eps_data = {
            "Min Before": tuple(sub_df["Min_Epsilon_Before"]),
            "Max Before": tuple(sub_df["Max_Epsilon_Before"]),
            "Min After": tuple(sub_df["Min_Epsilon_After"]),
            "Max After": tuple(sub_df["Max_Epsilon_After"])
        }
        w_eps = total_width / len(eps_data)
        for i, (label, vals) in enumerate(eps_data.items()):
            pos = x + (i * w_eps) - (total_width/2) + (w_eps/2)
            rects = ax[index, 0].bar(pos, vals, w_eps, label=label)
            ax[index, 0].bar_label(rects, padding=3, fmt='%.3f', fontsize=7, rotation=90)
        ax[index, 0].set_title(f"Epsilon (Depth {depth})")
        ax[index, 0].legend(loc='upper right', ncol=2, fontsize='small')

        # --- SUBPLOT 1: H Values (Log Scale) ---
        h_data = {
            "Max h Before": tuple(sub_df["Max_h_Before"]),
            # "Min h Before": tuple(sub_df["Min_h_Before"]),
            "Max h After": tuple(sub_df["Max_h_After"]),
            # "Min h After": tuple(sub_df["Min_h_After"])
        }
        w_h = total_width / len(h_data)
        h_colors = ['blue', 'green', 'red', 'black']
        for i, (label, vals) in enumerate(h_data.items()):
            pos = x + (i * w_h) - (total_width/2) + (w_h/2)
            rects = ax[index, 1].bar(pos, vals, w_h, label=label, color=h_colors[i])
            ax[index, 1].bar_label(rects, padding=3, fmt='%.3f', fontsize=7, rotation=90)
        # ax[index, 1].set_yscale('log')
        ax[index, 1].set_title(f"H Values - Log (Depth {depth})")
        ax[index, 1].legend(loc='upper right', ncol=2, fontsize='small')

        # --- SUBPLOT 2: Leaves (New Subplot) ---
        leaves_data = {
            "Leaves Before": tuple(sub_df["leaves_before"]),
            "Leaves After": tuple(sub_df["leaves_after"])
        }
        w_l = total_width / len(leaves_data)
        l_colors = ['purple', 'orange']
        for i, (label, vals) in enumerate(leaves_data.items()):
            pos = x + (i * w_l) - (total_width/2) + (w_l/2)
            rects = ax[index, 2].bar(pos, vals, w_l, label=label, color=l_colors[i])
            # Leaves are usually integers, but using .3f as requested
            ax[index, 2].bar_label(rects, padding=3, fmt='%.3f', fontsize=8, rotation=0)
            
        ax[index, 2].set_title(f"Leaves (Depth {depth})")
        ax[index, 2].legend(loc='upper right')

        # Common Formatting for all 3 subplots in the row
        for col in range(3):
            ax[index, col].set_xticks(x)
            ax[index, col].set_xticklabels(alphas)
            ax[index, col].set_xlabel("Alpha")
            ax[index, col].margins(y=0.25)

    plt.tight_layout()
    plt.show()

# Execution
# systems = df["system_name"].unique()
# for system in systems[3:4]:
#     plot_temp(df[df["system_name"] == system].copy())
plot_temp(df[df["system_name"] == "Jet"].copy())
```
